Introduction To Mathplotlib

```python
import matplotlib.pyplot as plt
import pandas as pd
```

Bar Plot

```python
course_enroll={
    "courses":["java","python","c","c++"],
    "enrolled":[30,50,90,60]
}
course_df=pd.DataFrame(course_enroll)
course_df
plt.Figure(figsize=(9,5))
plt.bar(course_df["courses"],course_df["enrolled"],color=["#6C63FF", "#00C9A7", "#FF6B6B", "#FFC75F"]

)
plt .title("Course Enrolled Student Overview",fontsize=15,fontweight='bold',pad=5)
plt.xlabel("Course",fontsize=12)
plt.ylabel("Enrolled students",fontsize=12)
plt.grid(axis='y',linestyle='--',alpha=0.3)
plt.gca().spines['right'].set_visible(False)
plt.tight_layout()

plt.show()
```

Line plot

```python
weekly_sales={
    "days":["Mon","tue","wed","thu","fri","sat","sun"],
    "first_week":[30000,40000,56000,20000,440000,20000,990000],
    "secound_week":[32000,50000,12000,36000,490000,40000,500000]
    }

sales=pd.DataFrame(weekly_sales)
plt.Figure(figsize=(9,5))
plt.plot(sales['days'],sales['first_week'],color='red',ls='--',marker='d',label="First week sales")
plt.plot(sales['days'],sales['secound_week'],color='blue',ls='--',marker='d',label="secound week sales")
plt.xlabel("day of week",fontsize=12)
plt.ylabel("sales amount per day",fontsize=12)
plt.legend()
plt.tight_layout()
plt.show()

```

Scatter Plot

```python
student_age={
    "sid":["s01","s02","s03","s04","s05","s06","s07","s08","s09","s10","s11"],
    "age":[20,22,21,22,23,29,26,28,20,30,40]
}
std_df=pd.DataFrame(student_age)
plt.Figure(figsize=(9,5))

plt.scatter(student_age['sid'],student_age['age'],color='green')
```

```python
plt.figure(figsize=(9,5))
plt.scatter(std_df['sid'], std_df['age'], color='green')
plt.show()
```

```python
plt.scatter(std_df['sid'], std_df['age'], color='red')
plt.show()
```

```python
plt.scatter(std_df['sid'], std_df['age'], s=150)
plt.show()
```

```python
plt.scatter(std_df['sid'], std_df['age'])
plt.title("Student Age Scatter Plot")
plt.show()
```

```python
plt.scatter(std_df['sid'], std_df['age'])
plt.xlabel("Student ID")
plt.ylabel("Age")
plt.show()
```

```python
plt.figure(figsize=(9,5))
plt.scatter(std_df['sid'], std_df['age'], color='blue')
plt.show()
```

Display all student ages with different colours.

```python
colors = ['red','blue','green','orange','purple',
          'pink','brown','cyan','yellow','black','gray']

plt.scatter(std_df['sid'], std_df['age'], c=colors)
plt.show()
```
