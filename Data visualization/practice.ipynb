```python
std_df = pd.read_csv("../pandas/student.csv")
print(std_df.head())
```

```python
data = std_df.select_dtypes(include='number')

plt.scatter(data.iloc[:, 0], data.iloc[:, 1], color='green')
plt.title("Data Analysis")
plt.xlabel(data.columns[0])
plt.ylabel(data.columns[1])
plt.grid(True)
plt.show()
```

```python
plt.plot(data.iloc[:, 0], data.iloc[:, 1], marker='o')
plt.title("Line Chart Analysis")
plt.xlabel(data.columns[0])
plt.ylabel(data.columns[1])
plt.grid(True)
plt.show()
```

```python
plt.bar(data.iloc[:, 0], data.iloc[:, 1], color='orange')
plt.title("Bar Chart Analysis")
plt.xlabel(data.columns[0])
plt.ylabel(data.columns[1])
plt.show()
```

```python
plt.hist(data.iloc[:, 0], bins=5, color='purple', edgecolor='black')
plt.title("Data Distribution")
plt.xlabel(data.columns[0])
plt.ylabel("Frequency")
plt.show()
```

```python
plt.barh(data.iloc[:, 0], data.iloc[:, 1], color='teal')
plt.title("Horizontal Bar Chart")
plt.xlabel(data.columns[1])
plt.ylabel(data.columns[0])
plt.show()
```

```python
colors = ['red', 'blue', 'green', 'orange', 'purple']
plt.bar(data.iloc[:, 0], data.iloc[:, 1],
        color=[colors[i % 5] for i in range(len(data))])
plt.title("Multi Colour Bar Chart")
plt.show()
```

```python
plt.boxplot(data.iloc[:, 1].dropna())
plt.title("Box Plot Analysis")
plt.ylabel(data.columns[1])
plt.grid(True)
plt.show()
```

```python
values = data.iloc[:, 1].dropna().head(5)
plt.pie(values, labels=values.index, autopct='%1.1f%%')
plt.title("Percentage Distribution")
plt.show()
```

```python
plt.plot(data.iloc[:, 0], label=data.columns[0])
plt.plot(data.iloc[:, 1], label=data.columns[1])
plt.title("Column Comparison")
plt.legend()
plt.show()
```

```python
plt.fill_between(data.index, data.iloc[:, 1], alpha=0.5)
plt.title("Area Chart")
plt.xlabel("Index")
plt.ylabel(data.columns[1])
plt.show()
```
