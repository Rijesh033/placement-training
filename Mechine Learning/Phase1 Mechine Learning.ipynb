```python
numerical_col=['age', 'bmi', 'children', 'charges']
for col in numerical_col:
    plt.figure(figsize=(6,4))
    sns.boxplot(df_insc[col])
    sns.histplot(df_insc[col], kde=True, color='orange', alpha=0.5)
    plt.title(f'Boxplot of {col}')
    plt.show()
```
