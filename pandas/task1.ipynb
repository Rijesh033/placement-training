```python

import pandas as pd

employees = pd.DataFrame({
    "name": ["Ravi", "Priya", "Rahul", "Anu", "Kiran"],
    "department": ["CSE", "ISE", "CSE", "ECE", "CSE"]
})

result = employees["department"].value_counts()

print(result)
```

Merge student details with their marks

```python

import pandas as pd

students = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["Ravi", "Priya", "Rahul"]
})

marks = pd.DataFrame({
    "id": [1, 2, 3],
    "marks": [85, 90, 78]
})

result = pd.merge(
    students,
    marks,
    on="id",
    how="inner"
)

print(result)
```

Find employees earning more than ₹30,000

```python

import pandas as pd

employees = pd.DataFrame({
    "name": ["Ravi", "Priya", "Rahul", "Anu", "Kiran"],
    "salary": [25000, 45000, 30000, 50000, 35000]
})

result = employees[employees["salary"] > 30000]

print(result)

```

Calculate average marks by subject

```python

import pandas as pd

data = pd.DataFrame({
    "student": ["Ravi", "Priya", "Rahul", "Anu", "Kiran"],
    "subject": ["Python", "Java", "Python", "Java", "Python"],
    "marks": [80, 90, 70, 85, 95]
})

result = data.groupby("subject")["marks"].mean()

print(result)
```

Find duplicate values

```python

import pandas as pd

data = pd.DataFrame({
    "name": ["Ravi", "Priya", "Ravi", "Anu", "Priya"],
    "marks": [80, 90, 80, 75, 90]
})

result = data[data.duplicated()]

print(result)
```

Find the second-highest salary

```python

import pandas as pd

df = pd.DataFrame({
    "name": ["Ravi", "Priya", "Rahul", "Anu", "Kiran"],
    "salary": [30000, 50000, 40000, 60000, 45000]
})

result = df.nlargest(2, "salary").iloc[1]

print(result)
```
