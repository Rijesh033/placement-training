compare,pivioting and melting

```python
import pandas as pd
```

Compare

```python
shop_7Sep={
    "Fruites":["Mango","Apple","Banana","Papaya"],
    "Prices":[100,150,60,30],
    "Quantity":[10,8,12,14]
}

stock_7Sep=pd.DataFrame(shop_7Sep)

shop_8Sep={
    "Fruites":["Mango","Apple","Banana","Papaya"],
    "Prices":[120,200,60,35],
    "Quantity":[15,6,24,17]
}

stock_8Sep=pd.DataFrame(shop_8Sep)

stock_7Sep.compare(stock_8Sep)
stock_7Sep.compare(stock_8Sep,align_axis=0)
stock_7Sep.compare(stock_8Sep,keep_shape=True)
stock_7Sep.compare(stock_8Sep,keep_equal=True)
stock_7Sep.compare(stock_8Sep,align_axis=0,keep_equal=True,keep_shape=True)
```

Pivoting

```python
monthly_sales={
    "product":["laptop","mobile","laptop","mobile","laptop","laptop"],
    "months":["jan","jan","feb","feb","mar","april"],
    "price":[95000,25000,67000,12000,65000,75000]
}
sales=pd.DataFrame(monthly_sales)
sales
sales.pivot_table(
    columns="months",
    index="product",
    values="price",
    aggfunc="sum"
)

```

Melting


```python
shop_sales={
    "product":["laptop","mobile","mouse"],
    "jan":[45000,14000,140],
    "feb":[55000,23000,250],
    "mar":[66000,45000,540]
}
sales_df=pd.DataFrame(shop_sales)
sales_df.melt(
    id_vars="product",
    var_name="month",
    value_name="price"
)
```
