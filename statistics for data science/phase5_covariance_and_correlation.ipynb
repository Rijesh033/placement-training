covariance and correlation


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sales_df = pd.read_csv("monthly_sales.csv")
sales_df.head(2)
```

# filter january months data from sales_df

```python
sales_month = sales_df[sales_df['Sales Month'] == 'Jan']
sales_month.tail(10)
```

#average sales in jan

```python
sales_jan = sales_df[sales_df['Sales Month'] == 'Jan']
sales_jan['Sales'].mean()
```

```python
sales_month = sales_df[sales_df['Sales Month'] == 'Jan']

display(sales_month)

print("Average January Sales:")
print(sales_month['Sales'].mean())
```

February month sales

```python
sales_feb = sales_df[sales_df['Sales Month'] == 'Feb']
display(sales_feb)
print("Average February Sales:")
print(sales_feb['Sales'].mean())
```

```python
sales_feb = sales_df[sales_df['Sales Month'] == 'Feb']
display(sales_feb)
```

```python
sales_feb[['Sales', 'Advertising']].corr()
```

```python
sns.scatterplot(x='Advertising', y='Sales', data=sales_feb)
plt.title('February Sales Correlation')
plt.show()
```

```python
sales_Mar = sales_df[sales_df['Sales Month'] == 'Mar']
sales_Mar['Advertising'].corr(sales_Mar['Sales'])
```

```python
sns.scatterplot(x='Advertising', y='Sales', data=sales_Mar)
plt.title('March Sales Correlation')
plt.show()
```
