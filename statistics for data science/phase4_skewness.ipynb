```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
student_marks={
    "s_id":["s01","s02","s03","s04","s05","s06","s07","s08","s09","s10","s11","s12","s13","s14","s15"],
    "marks":[45,67,78,89,90,56,45,67,78,89,90,56,45,67,78]
}
student_marks_df=pd.DataFrame(student_marks)
student_df=pd.DataFrame(student_marks_df)
student_df["marks"].skew()
```

```python
student_df["marks"].skew()
```

```python
sns.histplot(x="marks",data=student_df,bins=[i for i in range(40,100,5)])
plt.xticks(range(40,100,5))
plt.show()
```
