Mean

```python

import pandas as pd

titanicdf = pd.read_csv("titanic.csv")

print(titanicdf.columns.tolist())
```

```python

meanVal = titanicdf['fare'].mean()
print(meanVal)
```

```python

sns.histplot(data=titanicdf, x="fare", bins=15)
plt.axvline(meanVal, color="red")
plt.ylim(0, 400)
plt.show()
```

Median

```python
medianVal = np.median(titanicdf['fare'])
medianVal
```

```python

sns.histplot(x=titanicdf['fare'], bins=15)

plt.axvline(medianVal, color="green")
plt.ylim(0, 400)
plt.show()
```

Mode

```python
import statistics as st
modeVal=st.mode(titanicdf['fare'])
modeVal
```

```python

sns.histplot(x=titanicdf['fare'], data=titanicdf,
             bins=[i for i in range(0, 151, 10)])

plt.plot([modeVal, modeVal], [0, 351], c="yellow")
plt.ylim(0, 400)
plt.show()
```

Final out put

```python

sns.histplot(x=titanicdf['fare'], data=titanicdf, bins=[i for i in range(0, 151, 10)])
plt.plot([meanVal for i in range(351)], [i for i in range(351)], c="red", label="Mean")
plt.plot([medianVal for i in range(351)], [i for i in range(351)], c="green", label="Median")
plt.plot([modeVal for i in range(351)], [i for i in range(351)], c="yellow", label="Mode")

plt.legend()
plt.ylim(0, 400)
plt.title("Titanic Ticket Fare Central Tendency", fontweight="bold", fontsize=13)
plt.tight_layout()
plt.show()
```
